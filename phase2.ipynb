## Data Cleaning and Imputation (Phase 2 - Student 3)

In this section, data preprocessing was carried out to ensure the dataset is clean, structured, and ready for further modeling:

* **Initial Data Inspection:** Applied `df.head()` to briefly review the structure and first few records of the raw dataset before making any adjustments.
* **Handling Missing Values:** Addressed missing data by imputing missing values in numerical columns (`Age`) using the median, while replacing missing values in categorical variables (`Embarked`) with the most frequent value (mode).
* **Outlier Filtering:** Utilized the Interquartile Range (IQR) technique on the `Fare` column to set an upper threshold and remove extreme values, thereby improving overall data consistency.


```python
# Display initial raw data records
df.head()

# Fill missing values in 'Age' with median and 'Embarked' with mode
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Outlier detection and removal using IQR method for 'Fare'
q1_fare = df['Fare'].quantile(0.25)
q3_fare = df['Fare'].quantile(0.75)
iqr_fare = q3_fare - q1_fare
max_limit = q3_fare + 1.5 * iqr_fare

# Keep records within the valid range
df = df[df['Fare'] <= max_limit]

```
